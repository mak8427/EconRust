```python
import polars as pl
import pandas as pd
import plotly.express as px
from plotly.subplots import make_subplots
import plotly.graph_objects as go
```

```python
df= pl.read_csv('simulation_data.csv')
df = df.with_columns((df["Total Q_sold"] - df["Total Q_bought"]).alias("supply diff"))
df
```

```python
fig = make_subplots(rows=2, cols=1, shared_xaxes=True, vertical_spacing=0.1, subplot_titles=("Total Population", "Total Actors Money"))
fig.add_trace(go.Scatter(x=df["Day"],y=df["Total Population"]),row=1, col=1)
fig.add_trace(go.Scatter(x=df["Day"],y=df["Total Actors Money"]),row=2, col=1)

fig.update_layout(width=1200, height=800)

fig.show()
```

```python
fig = px.line(df, x="Day", y=["Total Q_bought","Total Q_sold"], title="Total Population")
fig.update_layout(width=1200, height=800)

fig.show()
```

```python
fig.show()
fig = px.line(df, x="Day", y='supply diff', title="supply diff")
fig.update_layout(width=1200, height=800)

fig.show()
```
